```python
import pandas as pd

df = pd.read_csv('202510-divvy-tripdata.csv')

df.drop(columns=['start_lat', 'start_lng', 'end_lat', 'end_lng'], inplace=True)

# 1. Ensure your date columns are in datetime format
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])

# 2. Calculate ride length in minutes
df['ride_length_mins'] = (df['ended_at'] - df['started_at']).dt.total_seconds() / 60

# 3. Extract the day of the week (e.g., Monday, Tuesday)
df['day_of_week'] = df['started_at'].dt.day_name()

# Keep only trips that lasted longer than 0 seconds (or 1 minute)
df = df[df['ride_length_mins'] > 0]

# Keep trips between 1 minute and 24 hours
df = df[(df['ride_length_mins'] >= 1) & (df['ride_length_mins'] <= 1440)]

# Drop rows where station names are missing
df = df.dropna(subset=['start_station_name', 'end_station_name'])

# Standardise rider types just in case
df['member_casual'] = df['member_casual'].replace({'Subscriber': 'member', 'Customer': 'casual'})

# (Optional) Check unique values to see if you need to fix them
print(df['member_casual'].unique())
print(df['rideable_type'].unique())

# Double-check for duplicates based on ride_id
df = df.drop_duplicates(subset=['ride_id'])

# Save the cleaned DataFrame to a new CSV file
df.to_csv('202510_cleaned.csv', index=False)
```
